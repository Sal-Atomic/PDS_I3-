Problem 3. Temperature Converter

```python
#Temperature Converter
def celsius_to_fahrenheit(c, round_to=1):
  F = (c*9/5+32)
  return round(F, round_to)
def fahrenheit_to_celsius(f, round_to=1):
  C = (f - 32)*5/9
  return round(C, round_to)

#(a) Use a for loop with range(0, 101, 20) to print the

print(" °C  |    °F") 
print("-----+-------")
for c in range (0, 101, 20):
  f = celsius_to_fahrenheit(c)
  print(f"{c:3}  |  {f:6.1f}")

#(b) convert 98.6F to Celsius
c= fahrenheit_to_celsius(98.6)
print(f"\n98.6 F = {c} C")

#(c) convert 37 C to Fahrenheit and back
f= celsius_to_fahrenheit(37)
c= fahrenheit_to_celsius(f)
print(f"37 C = {f} F")
print(f"{f} F = {c} C")

#(d) Round to 3 decimal places
print(celsius_to_fahrenheit(37.123, round_to=3))
```

Problem 4. Type-Conversion Detective

Part B. Write a function safe_int(text) that turns a string into an int. It must accept spaces around the number, negative numbers and decimals ("3.9" → 3), and return None when the text is not a 
number. 

```python
def safe_int(text):
  try:
    return int(float(text.strip()))
  except (ValueError, OverflowError, AttributeError):
    return None

for sample in ["42", "-8", "3.9", "abc", ""]:
  print(f"{sample}", f"{safe_int(sample)}", sep="->")
```

Problem 5. Letter Grade

or above 100 must return "Invalid".

Problem 6. Leap years and Days in a Month

A year is a leap year if it is divisible by 4, except century years (divisible by 100), which are leap years 
only if they are also divisible by 400. 

```python
#(a) Write is_leap(year) that returns True or False using one boolean expression with and / or. 
def is_leap(year):
  return(year%4==0 and year%100!=0) or (year%400==0)
#(b) Write days_in_month(month, year) that returns 28, 29, 30 or 31, or None if the month is not between 1 and 12. Use in to test the 30-day months (4, 6, 9, 11). 
def days_in_month(month, year):
  if month < 1 or month>12:
    return None
  if month==2:
    return 29 if is_leap(year) else 28
  if month in (4, 6, 9, 11):
    return 30
  else:
    return 31

print(f"Feb 2024: {days_in_month(2, 2024)} |" 
      f"Apr: {days_in_month(2, 2026)} |"
      f"month 13: {days_in_month(13, 2026)}"
      )
#(c) Print whether 1900, 2000, 2024 and 2026 are leap years.
for y in [1900, 2000, 2024, 2026]:
  print(y, "leap" if is_leap(y) else "not leap")
#(d) Count the leap years from 2000 to 2100 (both included) with a loop.
count = 0
for i in range(2000, 2101):
  if is_leap(i):
    count+=1  
print(f"Leap years from 2000 to 2100: {count}")
```

Problem 7: Digit Tools

Without converting numbers to strings, write three functions that work digit by digit with a while 
loop: 
- digit_sum(n) adds up the digits: 2026 → 2 + 0 + 2 + 6 = 10. 
- reverse_number(n) reverses the digits: 2026 → 6202. 
- is_palindrome(n) returns True when a number reads the same in both directions (12321). 

```python
def digit_sum(n):
  total =0
  while n>0:
    total += n%10
    n//=10
  return total

def reverse_number(n):
  reverse = 0
  while n > 0:
    reverse = reverse * 10 +n % 10
    n //=10
  return reverse

def is_palindrome(n):
  return n == reverse_number(n)
#(a) Test the three functions with 2026, 12321 and 9875.
for i in [2026, 12321, 9875]:
  print(i, f": digit sum = {digit_sum(i)}", f"reversed = {reverse_number(i)}", f"palindrome = {is_palindrome(i)}\n")
#(b) Count how many palindromes there are from 10 to 999. 
count = 0
for i in range(10, 1000):
  if is_palindrome(i):
    count += 1
print(f"Palindromes from 10 to 999: {count}")
```

Problem 8: Flexible Function: *args, **kwargs and lambda

```python
#(a) Write describe(*numbers, decimals=2, label="Data") that accepts any number of values, prints their count, min, max and mean (rounded to decimals), and returns them as a tuple. With no values it prints "<label>: no values" and returns None. 
def describe(*numbers, decimals=2, label="Data"):
  if not numbers:
    print(f"{label}: no values")
    return None

  count = len(numbers)
  minimum = min(numbers)
  maximum = max(numbers)
  mean = sum(numbers) / count

  print(
    f"{label}: n={count} min={minimum}"
    f"max={maximum} mean={mean:.{decimals}f}"
  )

  return count, minimum, maximum, round(mean, decimals)

#(b) Write make_tag(text, **style) that builds an HTML-like string from keyword arguments: make_tag("Hello", color="red", size=14) returns <span color="red" size="14">Hello</span>. 
def make_tag(text, **style):
  attributes = " ".join(
    f'{key}="{value}"' for key, value in style.items()
  )
  return f"<span {attributes}>{text}</span>"

#(c) Write two lambdas: square (returns x²) and is_adult (returns True when age ≥ 18)
square = lambda x: x ** 2

is_adult = lambda age: age >= 18

describe(12, 7, 30, 4)
describe(3.12159, 2.71828, decimals=3, label="Constants")
describe()

print(make_tag("Hello", color="red", size=14))

print(square(9), is_adult(17), is_adult(21))
```

Problem 9. Electricity Bill with Progressive Tiers

An electricity company charges each kWh according to the tier it falls in, like income tax. A fixed 
service fee of 2,000 riel is added, then 10 % VAT is applied to the total.

```python
def electricity_bill(kwh):
  if kwh < 0:
    return None

  #Calculate consumption in each tier
  tier1 = min(kwh, 50)
  tier2 = min(max(kwh - 50, 0), 150)
  tier3 = max(kwh - 200, 0)

  #Calculate the electricity cost
  cost = tier1 * 480 +tier2 * 610 +tier3 * 730

  #Add service fee and VAT
  total = (cost + 2000) * 1.10

  return round(total)

#(a) 
print(electricity_bill(120))

#(b)
consumptions = [0, 35, 50, 120, 200, 350]
for i in consumptions:
  bill = electricity_bill(i)
  usd = bill/4100

  print(f"{i:3} i -> {bill:>10,} riel ({usd:7.2f} USD)")

#(c)
kwh = float(input("\nEnter your monthly electricity consumption (kwh): "))

bill = electricity_bill(kwh)

if bill is None:
  print("Invalid consumption, please enter a non-negative value.")
else:
  usd = bill / 4100
  print(f"Your electricity bill: {bill:,} riel ({usd:.2f} USD)")
```

Problem 10. Prime Numbers and Goldbach's Conjecture

```python
#(a)
def is_prime(n):
  if n < 2:
    return False

  d = 2
  while d * d <= n:
    if n % d == 0:
      return False
    d += 1
  return True

#(b)
print("Primes below 50:", end=" ")

for n in range(2, 50):
  if is_prime(n):
    print(n, end=" ")

print()

#Count primes below 100
count = 0

for n in range(2, 100):
  if is_prime(n):
    count += 1

print("Number of primes below 100:", count)

#(c)
def goldbach(n):
  for p in range(2, n):
    if is_prime(p):
      q = n - p

      if is_prime(q):
        return p, q

  return None

#Print Goldbach pairs for even numbers from 4 to 30
for n in range(4, 31, 2):
  pair = goldbach(n)
  p, q = pair
  print(f"{n} = {p} + {q}")

#(d) Check Goldbach's conjecture for every even number
holds = True

for n in range(4, 10001, 2):
  if goldbach(n) is None:
    holds = False
    break

print(
  f"Goldbach holds for every even number "
  f"from 4 to 10000: {holds}"
)
```
